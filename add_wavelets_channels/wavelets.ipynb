```julia
using Pkg
Pkg.activate(".")
using Wavelets
using Images
using PreprocessingImages
using TestImages
```

```julia
# obtém imagem de teste
img = testimage("mandril") |> x->imresize(x, ratio=1/2)
@show size(img);
```

```julia
# transforma imagem colorida em cinza e em float32
img_gray  = Gray.(img)
@show size(img_gray)
img_float = channelview(img_gray) .|> Float32
H, W = size(img_float)
mosaicview(img, img_gray, nrow=1)
```

```julia
# utiliza representação float32 para cálculo da transformação
wt = wavelet(WT.haar)
imgt = dwt(img_float, wt, 1)
@show size(imgt);
```

```julia
# esta transformação de haar em particular calcula 4 canais de resposta
# estes canais são dispostos em uma única matriz quadrada
half_h, half_w = size(imgt) .÷ 2
LL = imgt[1:half_h, 1:half_w]
LH = imgt[1:half_h, half_w+1:end]
HL = imgt[half_h+1:end, 1:half_w]
HH = imgt[half_h+1:end, half_w+1:end]
@show size(LL);
```

```julia
imgt  # contém números negativos
imgt_plot = wplotim(imgt,2) .|> Gray   # normaliza para visualização
```

```julia
# apresenta imagem original em cinza e transformações
img_gray_resized = imresize(img_gray, ratio=1/2)
img_LL = Gray.(wplotim(LL,2))
img_LH = Gray.(wplotim(LH,2))
img_HL = Gray.(wplotim(HL,2))
img_HH = Gray.(wplotim(HH,2))

mosaicview(img_gray_resized, img_LL, img_LH, img_HL, img_HH, nrow=1)
```
