```julia
using Pkg
envpath = expanduser("../")
Pkg.activate(envpath)

# libs
using Images
using DataFrames
using CSV
using PascalVocTools; pv=PascalVocTools
```

```julia
# input folders
datasetPath = "~/tests/pascalvoc/dataset/pascalvoc/VOCdevkit/VOC2012/"
imagesPath  = datasetPath * "JPEGImages/"
masksPath   = datasetPath * "SegmentationClass/"

trainlist = datasetPath * "ImageSets/Segmentation/train.txt"
testlist  = datasetPath * "ImageSets/Segmentation/val.txt"
```

```julia
lines = readlines(expanduser(trainlist))
dftrain = DataFrame(filename = lines)

lines = readlines(expanduser(testlist))
dftest = DataFrame(filename = lines)

size(dftrain), size(dftest)
```

```julia
dftrain.X = imagesPath .* dftrain.filename .* ".jpg"
dftest.X  = imagesPath .* dftest.filename .* ".jpg"

dftrain.y = masksPath .* dftrain.filename .* ".png"
dftest.y  = masksPath .* dftest.filename .* ".png"

select!(dftrain, Not(:filename))
select!(dftest, Not(:filename))

size(dftrain), size(dftest)
```

```julia
# save datasets
CSV.write("dftrain.csv", dftrain)
CSV.write("dftest.csv", dftest)
```
